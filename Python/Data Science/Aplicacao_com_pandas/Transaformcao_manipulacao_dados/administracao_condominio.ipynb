# Adiministação de Condominio

 administração de condomínios é uma tarefa que requer muita atenção e organização. Entre as diversas responsabilidades da gestão está o recebimento dos aluguéis dos locatários. Para garantir uma boa saúde financeira do empreendimento, é fundamental que esses pagamentos sejam feitos de forma regular e pontual. No entanto, sabemos que nem sempre isso acontece.

Pensando nisso, proponho um desafio de tratamento de dados com o objetivo de analisar o atraso no pagamento de aluguéis no condomínio de alguns moradores. Disponibilizo a base de dados dados_locacao_imoveis.json, que contém informações sobre o apartamento dos locatários, o dia acordado para o pagamento do aluguel, o dia da efetivação do pagamento de aluguel e o valor do aluguel.

Com essas informações, o desafio do projeto 2: administração de condomínios será similar ao desafio do projeto 1, abrir a base de dados com Pandas e aplicar o json_normalize no DataFrame.


```python
import pandas as pd
import requests
import numpy as np 
```

```python

url = 'https://cdn3.gnarususercontent.com.br/2928-transformacao-manipulacao-dados/dados_locacao_imoveis.json'

# Simular um navegador
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/123.0.0.0 Safari/537.36"
}

# Fazer a requisição
response = requests.get(url, headers=headers)

# Verificar se o acesso foi permitido
if response.status_code == 200:
    data = response.json()
    dados_adm_condominios = pd.DataFrame(data)
    print(dados_adm_condominios.head())
else:
    print(f"Erro {response.status_code}: Acesso negado")
```

```python
dados_adm_condominios = pd.json_normalize(dados_adm_condominios['dados_locacao'])
dados_adm_condominios
```

Lemos a base de dados no desafio anterior, agora podemos avançar nas transformações desses dados. Então, da mesma forma que o projeto 1, o desafio do projeto 2 está listado em algumas metas:

- Remover os dados em listas dentro do DataFrame;
- Verificar os tipos de dados;
- Identificar colunas numéricas;
- Transformar a coluna numérica para o tipo numérico.

```python
colunas = list(dados_adm_condominios.columns)
colunas
```

```python
dados_adm_condominios = dados_adm_condominios.explode(colunas[1:])
dados_adm_condominios
```

```python
dados_adm_condominios.reset_index(inplace=True,drop=True)
dados_adm_condominios.head()
```

```python
dados_adm_condominios['valor_aluguel'] = dados_adm_condominios['valor_aluguel'].apply(lambda x: x.replace('$','').replace(',','.').replace('reais','').strip())
dados_adm_condominios
```

```python
dados_adm_condominios['valor_aluguel'] = dados_adm_condominios['valor_aluguel'].astype(np.float64)
```

```python
dados_adm_condominios.info()
```

```python
dados_adm_condominios['valor_aluguel'] = dados_adm_condominios['valor_aluguel'].astype(np.float64)
```

Buscando explicar a organização da identificação dos apartamentos, durante a criação do conjunto de dados, foi adicionado o texto (blocoAP). Esse texto informa que os nomes dos apartamentos estão organizados com a letra do bloco seguida do número do apartamento. No entanto, isso não traz nenhuma informação para nossos dados, sendo interessante realizar a remoção desse texto no conjunto de dados.

Com isso, manipule os textos na coluna apartamento para remover o texto (blocoAP) do DataFrame.



```python
dados_adm_condominios['apartamento'].head()
```

```python

dados_adm_condominios['apartamento'] = dados_adm_condominios['apartamento'].str.replace('\(blocoAP\)','',regex=True)

dados_adm_condominios['apartamento'].head()
```

Nas colunas datas_de_pagamento e datas_combinadas_pagamento, temos datas em formato 'dia/mês/ano' (dd/mm/AAAA). Transforme esses dados para o tipo datetime e busque uma forma de visualização de subconjunto que possa contribuir no objetivo do contexto que os dados estão inseridos.

```python
dados_adm_condominios.head()
```

```python
dados_adm_condominios['datas_combinadas_pagamento'] = pd.to_datetime(dados_adm_condominios['datas_combinadas_pagamento'], format='%d/%m/%Y')
dados_adm_condominios['datas_de_pagamento'] = pd.to_datetime(dados_adm_condominios['datas_de_pagamento'], format='%d/%m/%Y')
```

```python
# Para contribuir na solução do contexto é possível calcular a diferença de dias
# entre a data combinada e a data de pagamento com dt.days
dados_adm_condominios['atraso'] = (dados_adm_condominios['datas_de_pagamento'] - dados_adm_condominios['datas_combinadas_pagamento']).dt.days
```

```python
# Calcular a média de tempo de atraso por apartamentos
media_atraso = dados_adm_condominios.groupby(['apartamento'])['atraso'].mean()

# Visualizar o resultado
media_atraso
```
